```python
# step1 
days=['Mon','Tue',"Wed","Thu","Fri","Sat","Sun"]
temps = [31, 33, 29, 34, 35, 32, 30]
print("First Days : ",days[0])
```

```python
print("Day of the weeks : ",len(temps))
print("The last day of the week : ",days[6],temps[6])
```

```python
print("highest Temperature : ",max(temps))
print("Lowert Temperature : ",min(temps))
average=sum(temps)/len(temps)
print(average)
```

```python
hottest=temps.index(max(temps))
print("Hottest Day ",days[hottest])
```

```python
print("Sorted : ", sorted(temps))
print("Original : ",temps)
```
