```python
football_list = ["Dara", "Vibol", "Kosal", "Dara", "Bopha"]
football = set(football_list)
print("List size:", len(football_list), "| Set size:", len(football))
```

```python
music = {"Sophea", "Kosal", "Malis", "Dara"}

```

```python
print("Both clubs :", sorted(football & music)) # intersection and
print("Any club :", sorted(football | music)) # union or
print("Football only :", sorted(football - music)) # difference -
print("Exactly one :", sorted(football ^ music)) # symmetric difference
```

```python
music.update({"Chenda"})
music.discard("Malis")
print("Music now :", sorted(music))
print("Is Vibol in music?", "Vibol" in music)
```
