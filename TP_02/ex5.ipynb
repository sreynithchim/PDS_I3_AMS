```python
prices = {"apple": 0.50, "banana": 0.25, "mango": 1.00}
print("Price of mango:", prices["mango"])
```

```python
prices["orange"] = 0.75
prices["apple"] = 0.60
del prices["banana"]
print("Prices:", prices)
```

```python
print("Price of grape:", prices.get("grape", "not sold"))
```

```python
for fruit, price in prices.items():
    print(f"{fruit:<7} ${price:.2f}")
```

```python
basket = {"apple": 4, "mango": 2, "orange": 3}
total = 0
for fruit, qty in basket.items():
    total = total + prices[fruit] * qty
print(f"Total bill: ${total:.2f}")
```

```python
class Student:
    def __init__(self, id, name, gpa):
        self.id = id
        self.name = name
        self.gpa = gpa

    def show(self):
        print(self.id, self.name, self.gpa)

def update_gpa(s, new_gpa):     # receives a reference
    s.gpa = new_gpa

s1 = Student(1, 'Sokha', 3.5)
print(s1.name)                  # Sokha
update_gpa(s1, 4.0)
s1.show()
```

```python
def add(a, b):              # return function
    return a + b

def show_message():         # void function
    print('Hello')

print("Result : ",add(3, 5))            
show_message()
```

```python
my_list = ["Apple", 20, 3.5, "Banana"]
"""Output the Origal List"""
print("Original list:", my_list)

"""Add Element using append"""
my_list.append("Orange")
print("After adding:", my_list)

"""Remove Element"""
my_list.remove(20)
print("After removing:", my_list)

"""Sort list"""
numbers = [5, 2, 8, 1, 4]
numbers.sort()
print("After sorting:", numbers)
```
